```python


a = 1103515245
c = 12345
m = 2**31

x = 7

random_numbers = []

for i in range(100):
    x = (a * x + c) % m
    u = x / m
    random_numbers.append(u)

print("100 случайных чисел:")
print(random_numbers)
```

```python
import random
import math

lam = 4
numbers = []

for i in range(10):
    u = random.random()
    y = -math.log(1 - u) / lam
    numbers.append(y)

print("10 случайных чисел экспоненциального распределения:")
print(numbers)
```
